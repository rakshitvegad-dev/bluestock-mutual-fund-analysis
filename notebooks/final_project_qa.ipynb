```python
# ============================================================
# CELL 5 — NORMALIZED PERFORMANCE VALIDATION
# ============================================================

print("=" * 60)
print("NORMALIZED PERFORMANCE VALIDATION")
print("=" * 60)

# ------------------------------------------------------------
# 1. Basic information
# ------------------------------------------------------------

print("\n1. Dataset Information")

print("Shape:", normalized.shape)

print("Date range:")
print(
    normalized["date"].min(),
    "to",
    normalized["date"].max()
)

print(
    "Number of unique series:",
    normalized["Series"].nunique()
)

print("\nSeries:")
print(
    normalized["Series"].unique()
)

# ------------------------------------------------------------
# 2. Check starting value
# ------------------------------------------------------------

print("\n2. Starting Values")

first_date = normalized["date"].min()

starting_values = (
    normalized[
        normalized["date"] == first_date
    ]
    [
        ["Series", "Normalized_Value"]
    ]
)

display(starting_values)

# ------------------------------------------------------------
# 3. Check for invalid normalized values
# ------------------------------------------------------------

print("\n3. Invalid Values")

print(
    "Negative normalized values:",
    (
        normalized["Normalized_Value"] < 0
    ).sum()
)

print(
    "Missing normalized values:",
    normalized["Normalized_Value"].isna().sum()
)

print(
    "Infinite values:",
    np.isinf(
        normalized["Normalized_Value"]
    ).sum()
)

# ------------------------------------------------------------
# 4. Performance summary
# ------------------------------------------------------------

print("\n4. Final Performance")

performance_check = []

for series in normalized["Series"].unique():

    temp = normalized[
        normalized["Series"] == series
    ].sort_values("date")

    start_value = temp[
        "Normalized_Value"
    ].iloc[0]

    final_value = temp[
        "Normalized_Value"
    ].iloc[-1]

    total_return = (
        (final_value / start_value) - 1
    ) * 100

    performance_check.append({
        "Series": series,
        "Start_Value": start_value,
        "Final_Value": final_value,
        "Total_Return_pct": total_return
    })

performance_check = pd.DataFrame(
    performance_check
).sort_values(
    "Total_Return_pct",
    ascending=False
)

display(performance_check)

# ------------------------------------------------------------
# 5. Check benchmarks
# ------------------------------------------------------------

print("\n5. Benchmark Check")

required_benchmarks = [
    "NIFTY50",
    "NIFTY100"
]

for benchmark in required_benchmarks:

    if benchmark in normalized["Series"].unique():
        print(f"✅ {benchmark} present")
    else:
        print(f"❌ {benchmark} missing")

# ------------------------------------------------------------
# 6. Final normalized QA
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("NORMALIZED PERFORMANCE QA SUMMARY")
print("=" * 60)

checks = {
    "Dataset contains data":
        len(normalized) > 0,

    "NIFTY50 present":
        "NIFTY50" in normalized["Series"].unique(),

    "NIFTY100 present":
        "NIFTY100" in normalized["Series"].unique(),

    "No negative normalized values":
        (normalized["Normalized_Value"] < 0).sum() == 0,

    "No missing normalized values":
        normalized["Normalized_Value"].isna().sum() == 0,

    "No infinite values":
        np.isinf(
            normalized["Normalized_Value"]
        ).sum() == 0
}

for check, result in checks.items():

    if result:
        print(f"✅ {check}")
    else:
        print(f"❌ {check}")

print("=" * 60)
```

```python
# ============================================================
# CELL 6 — FINAL PROJECT CONSISTENCY CHECK
# ============================================================

print("=" * 70)
print("FINAL PROJECT CONSISTENCY CHECK")
print("=" * 70)

# ------------------------------------------------------------
# 1. Get Top 5 from final scorecard
# ------------------------------------------------------------

top5_scorecard = (
    scorecard
    .sort_values("Overall_Rank")
    .head(5)
    [["amfi_code", "scheme_name", "Overall_Rank"]]
    .copy()
)

print("\n1. TOP 5 FUNDS FROM FINAL SCORECARD")
display(top5_scorecard)

# ------------------------------------------------------------
# 2. Get funds from Tracking Error
# ------------------------------------------------------------

te_funds = set(
    tracking_error["scheme_name"]
    .dropna()
    .astype(str)
)

# ------------------------------------------------------------
# 3. Get funds from Rolling Tracking Difference
# ------------------------------------------------------------

rolling_funds = set(
    rolling_tracking["scheme_name"]
    .dropna()
    .astype(str)
)

# ------------------------------------------------------------
# 4. Get funds from normalized performance
# ------------------------------------------------------------

benchmark_names = {
    "NIFTY50",
    "NIFTY100"
}

normalized_funds = set(
    normalized["Series"]
    .dropna()
    .astype(str)
) - benchmark_names

# ------------------------------------------------------------
# 5. Scorecard Top 5 set
# ------------------------------------------------------------

scorecard_funds = set(
    top5_scorecard["scheme_name"]
    .astype(str)
)

# ------------------------------------------------------------
# 6. Compare all datasets
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TOP 5 CONSISTENCY")
print("=" * 70)

print(
    "Scorecard Top 5:",
    len(scorecard_funds)
)

print(
    "Tracking Error funds:",
    len(te_funds)
)

print(
    "Rolling Tracking funds:",
    len(rolling_funds)
)

print(
    "Normalized Performance funds:",
    len(normalized_funds)
)

# ------------------------------------------------------------
# 7. Compare Scorecard vs Tracking Error
# ------------------------------------------------------------

print("\n2. SCORECARD vs TRACKING ERROR")

missing_from_te = (
    scorecard_funds - te_funds
)

extra_in_te = (
    te_funds - scorecard_funds
)

if not missing_from_te and not extra_in_te:

    print(
        "✅ Top 5 funds match Tracking Error dataset"
    )

else:

    print(
        "⚠️ Difference detected"
    )

    if missing_from_te:
        print(
            "Missing from Tracking Error:"
        )
        print(missing_from_te)

    if extra_in_te:
        print(
            "Extra funds in Tracking Error:"
        )
        print(extra_in_te)

# ------------------------------------------------------------
# 8. Compare Scorecard vs Rolling Tracking
# ------------------------------------------------------------

print("\n3. SCORECARD vs ROLLING TRACKING")

missing_from_rolling = (
    scorecard_funds - rolling_funds
)

extra_in_rolling = (
    rolling_funds - scorecard_funds
)

if not missing_from_rolling and not extra_in_rolling:

    print(
        "✅ Top 5 funds match Rolling Tracking dataset"
    )

else:

    print(
        "⚠️ Difference detected"
    )

    if missing_from_rolling:
        print(
            "Missing from Rolling Tracking:"
        )
        print(missing_from_rolling)

    if extra_in_rolling:
        print(
            "Extra funds in Rolling Tracking:"
        )
        print(extra_in_rolling)

# ------------------------------------------------------------
# 9. Compare Scorecard vs Normalized Performance
# ------------------------------------------------------------

print("\n4. SCORECARD vs NORMALIZED PERFORMANCE")

missing_from_normalized = (
    scorecard_funds - normalized_funds
)

extra_in_normalized = (
    normalized_funds - scorecard_funds
)

if not missing_from_normalized and not extra_in_normalized:

    print(
        "✅ Top 5 funds match Normalized Performance dataset"
    )

else:

    print(
        "⚠️ Difference detected"
    )

    if missing_from_normalized:
        print(
            "Missing from Normalized Performance:"
        )
        print(missing_from_normalized)

    if extra_in_normalized:
        print(
            "Extra funds in Normalized Performance:"
        )
        print(extra_in_normalized)

# ------------------------------------------------------------
# 10. Benchmark consistency
# ------------------------------------------------------------

print("\n5. BENCHMARK CONSISTENCY")

required_benchmarks = {
    "NIFTY50",
    "NIFTY100"
}

rolling_benchmarks = set(
    rolling_tracking["Benchmark"]
    .dropna()
    .astype(str)
)

normalized_benchmarks = set(
    normalized["Series"]
    .dropna()
    .astype(str)
) & required_benchmarks

print(
    "Rolling Tracking benchmarks:",
    rolling_benchmarks
)

print(
    "Normalized benchmarks:",
    normalized_benchmarks
)

if required_benchmarks.issubset(
    rolling_benchmarks
):

    print(
        "✅ Rolling Tracking contains both benchmarks"
    )

else:

    print(
        "❌ Rolling Tracking benchmark issue"
    )

if required_benchmarks.issubset(
    normalized_benchmarks
):

    print(
        "✅ Normalized Performance contains both benchmarks"
    )

else:

    print(
        "❌ Normalized Performance benchmark issue"
    )

# ------------------------------------------------------------
# 11. Final QA result
# ------------------------------------------------------------

all_consistent = (
    scorecard_funds == te_funds
    and
    scorecard_funds == rolling_funds
    and
    scorecard_funds == normalized_funds
    and
    required_benchmarks.issubset(
        rolling_benchmarks
    )
    and
    required_benchmarks.issubset(
        normalized_benchmarks
    )
)

print("\n" + "=" * 70)
print("FINAL CONSISTENCY RESULT")
print("=" * 70)

if all_consistent:

    print(
        "🎉 PASS — All Top 5 funds and benchmarks "
        "are consistent across the project."
    )

else:

    print(
        "⚠️ REVIEW REQUIRED — Some datasets are not consistent."
    )

print("=" * 70)
```
